*Limiti dei dati: prezzi Amazon letti da grafici storici (±1-2 giorni), buchi dal 20/09 al 04/10/2025 per 15 e 16, solo 4 modelli.*

```python
# ricarica automaticamente i moduli di src/utils/ se li modifichi
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path

# trova la cartella src/ partendo dalla cartella corrente, qualunque sia la cartella di lavoro del kernel
radice = next(d for d in [Path.cwd(), *Path.cwd().parents] if (d / "src" / "utils" / "dati.py").exists())
sys.path.append(str(radice / "src"))

from utils.dati import carica_modello
from utils import calcoli, grafici

modelli = ["iPhone14", "iPhone15", "iPhone16", "iPhone17"]
dati = {m: carica_modello(m) for m in modelli}
```

## 1. Sconti nel tempo
Curva dello sconto per ogni modello (giorni dal lancio) con i trimestri T1-T8 sullo sfondo. Sotto, sconto medio per trimestre (heatmap) e distribuzione giornaliera (box plot, primo anno).

```python
grafici.sconto_per_giorno(dati).show()
```

```python
grafici.heatmap_trimestri(dati).show()
```

## 2. Quando conviene
Sconto nei primi 30/90/180/365 giorni: medio, massimo e % di giorni in cui il prezzo era sotto quello di lancio. Poi il giorno migliore del primo anno per ogni modello.

```python
calcoli.tabella_sconti(dati)
```

```python
minimi = calcoli.riepilogo_minimi(dati)
display(minimi)
grafici.minimi(minimi).show()
```

## 3. Costo dell'attesa
- **Spreco**: quanti € in più paghi comprando al giorno N rispetto al prezzo minimo dell'anno.
- **Guadagno dell'attesa**: quanti € risparmieresti aspettando il minimo successivo (con il senno di poi).

```python
curve = {m: calcoli.costo_attesa(df) for m, df in dati.items()}
grafici.costo_attesa(curve).show()
```

## 4. Stagionalità
Variazione % media del prezzo per mese e per settimana (Prime Day, Black Friday, Natale). Il primo periodo di ogni modello è escluso. Con 4 modelli sono indicazioni, non regole; i picchi positivi (es. settembre 2026 dell'iPhone 17) possono dipendere da scorte o venditori terzi.

```python
mensile = calcoli.stagionalita_mensile(dati)
settimanale = calcoli.stagionalita_settimanale(dati)
grafici.stagionalita(mensile, settimanale).show()
```

## 5. Offerte e volatilità
*Offerta* = periodo continuo con prezzo almeno 3% sotto la mediana degli ultimi 60 giorni. Variazioni/mese = quante volte cambia il prezzo.

```python
profilo = calcoli.profilo_offerte(dati)
display(profilo)
grafici.offerte(profilo).show()
```

## 6. Effetto del lancio del nuovo modello
Andamento del prezzo del modello precedente nei ±28 giorni intorno al lancio del successivo (settembre). Serve a capire se conviene comprare *prima* o *dopo* l'arrivo del nuovo modello.

```python
finestre = calcoli.effetto_lancio(dati)
grafici.effetto_lancio(finestre).show()
```
